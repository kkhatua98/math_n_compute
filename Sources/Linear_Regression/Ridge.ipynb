$$
\begin{aligned}
& \text { SSE } \\
& =\|\vec{y}-X \beta\|^2 \\
& =\|\vec{y}-X \hat{\beta}+X \hat{\beta}-X \beta\|^2 \\
& =[(\vec{y}-X \hat{\beta})+(X \hat{\beta}-X \beta)]^{\top}[(\vec{y}-X \hat{\beta})+(X \hat{\beta}-X \beta)] \\
& =\left[(\vec{y}-X \hat{\beta})^{\top}+(X \hat{\beta}-X \beta)^{\top}\right][(\vec{y}-X \hat{\beta})+(X \hat{\beta}-X \beta)] \\
& =(\vec{y}-X \hat{\beta})^{\top}(\vec{y}-X \hat{\beta})+(\vec{y}-X \hat{\beta})^{\top}(X \hat{\beta}-X \beta)+(X \hat{\beta}-X \beta)^T(\vec{y}-X \hat{\beta}) +(X \hat{\beta}-X \beta)^{\top}(X \hat{\beta}-X \beta) \\
& ={SSE}_{OLS}+(\vec{y}-X \hat{\beta})^{\top}(X \hat{\beta}-X \beta)+(X \hat{\beta}-X \beta)^T(\vec{y}-X \hat{\beta})+(\hat{\beta}-\beta)^{\top} X^{\top} X(\hat{\beta}-\beta)
\end{aligned}
$$

Calculate (1)

$$
\begin{aligned}
& (\vec{y}-X \hat{\beta})^{\top}(X \hat{\beta}-X \beta) \\
&= \left(\vec{y}^{\top}-\hat{\beta}^{\top} X^{\top}\right)(X \hat{\beta}-X \beta) \\
&= \vec{y}^{\top} X \hat{\beta}-\vec{y}^{\top} X \beta-\hat{\beta}^{\top} X^{\top} X \hat{\beta}+\hat{\beta}^{\top} X^{\top} X \beta \\
&= \vec{y}^{\top} X \hat{\beta}-\vec{y}^{\top} X \beta-\hat{\beta}^{\top} X^{\top} \vec{y}+\vec{y}^{\top} X \beta \\
&= \vec{y}^{\top} X \hat{\beta}-\hat{\beta}^{\top} X^{\top} \vec{y} \\
&= \hat{\beta}^{\top} X^{\top} \vec{y}-\hat{\beta}^{\top} X^{\top} \vec{y} \\

\end{aligned}
$$




Calculate (2)


$$
\begin{aligned}

\end{aligned}
$$

Now $X^TX$ is a symmetric matrix, so there is a $P$ and $\Lambda$ such that

$$X^TX=P\Lambda P^T$$
where,
$$\Lambda = \begin{bmatrix}\lambda \ \ 0 \\
0 \ \ \mu \end{bmatrix}, P=(\vec{u} \ \ \vec{v})$$


here, $\lambda$ and $\mu$ are eigenvalues and $\vec{u},\vec{v}$ are the eigenvectors of $X^TX$

$$
\begin{aligned}
&SSE \\
=&{SSE}_{OLS}+(\hat{\beta}-\beta)^{\top} X^{\top} X(\hat{\beta}-\beta) \\ 
=&{SSE}_{OLS}+(\hat{\beta}-\beta)^{\top} P\Lambda P^T(\hat{\beta}-\beta) \\ 
=&{SSE}_{OLS}+(\beta - \hat{\beta})^{\top} P\Lambda P^T(\beta-\hat{\beta}) \\
=&{SSE}_{OLS}+(\gamma - \hat{\gamma})^T \Lambda (\gamma - \hat{\gamma}) \ \ 
\end{aligned}
$$

In the above equation, $P^T\beta = \gamma$

$(\gamma - \hat{\gamma})^T \Lambda (\gamma - \hat{\gamma})$ is equation of an ellipse with center at $\hat{\gamma}=P^T\hat{\beta}$ and axes dentoed by the equations $\gamma=\hat{\gamma} \ \ ie. \ \  P^T\beta=P^T\hat{\beta}$
